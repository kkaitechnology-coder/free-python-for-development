## Virtual Environments and Dependency Isolation

Managing dependencies is one of the first practical skills every Python developer must master. Installing packages globally with **pip install** quickly leads to version conflicts between projects — a situation developers call dependency hell.

A **virtual environment** solves this by giving each project its own isolated Python interpreter and package directory. Changes made inside one environment never affect another project or the system Python.

> 💡 **Key Mental Model** Think of each project as a self-contained workspace. The libraries living in Project A’s environment are invisible to Project B. This isolation is what makes reproducible builds and clean collaboration possible.

### How venv Works

- Running `python -m venv .venv` creates a new folder (commonly named `.venv`).
- Inside that folder you will find a `pyvenv.cfg` file and symlinks (or copies) to the base Python interpreter.
- Activating the environment adjusts `PATH` so that `python` and `pip` now point to the isolated copies.
- All packages installed with `pip` go into the environment’s own `site-packages` directory.

**Classic venv**

- Built into the standard library
- Lightweight and universal
- Requires manual activation & requirements.txt

**Modern Tools (Poetry / uv)**

- Declarative dependency specification
- Automatic lock files for exact versions
- Faster resolution and installation
- Built-in virtual-environment management

In daily practice most teams now prefer Poetry or the extremely fast **uv** because they eliminate the need to remember activation commands and guarantee that every developer and every CI runner uses identical package versions.

## Script Execution Lifecycle, Entry Points & Module Resolution

When you run a Python file the interpreter reads it from top to bottom and executes every statement in order. Understanding this linear flow — and the special role of the `if __name__ == "__main__"` guard — is essential for writing code that can be both executed as a script and imported as a reusable module.

```python
# greetings.py
def say_hello(name: str) -> str:
    """Return a friendly greeting."""
    return f"Hello, {name}!"

def main() -> None:
    # This block runs only when the file is executed directly
    user = input("Your name: ")
    print(say_hello(user))

if __name__ == "__main__":
    main()
```

**Execution Rules**

- Top-to-bottom statement execution.
- `__name__` is set to `"__main__"` only when the file is the entry point.
- When imported, `__name__` becomes the module name, so the `main()` call is skipped.
- This pattern lets you keep demo / test code without polluting library usage.

### Module Resolution Internals

Importing is not a simple file open. Python first consults the dictionary `sys.modules`. If the module is already present, the cached object is returned immediately — avoiding repeated disk I/O and re-execution.

When the module is not yet loaded, the interpreter searches the list of directories stored in `sys.path`. That list always contains the current working directory, the standard-library locations, and any paths added by site-packages or virtual environments.

> 💡 **Practical Tip** Never mutate `sys.path` at runtime in production code. Prefer proper package installation or the `PYTHONPATH` environment variable so that import behaviour stays predictable across machines.

## Interactive Debugging, Profiling & Production Tooling

Professional Python development is rarely a pure “write → run” loop. Modern workflows emphasise rapid interactive exploration, precise debugging, performance measurement, and strict production hygiene.

### Interactive Exploration & Debugging

Tools such as **IPython** and the Rich-enhanced REPL let you treat the interpreter like a scientific calculator: experiment with snippets, inspect objects, and see beautifully formatted results instantly.

```python
# Using the built-in breakpoint (Python 3.7+)
def calculate_discount(price: float, rate: float) -> float:
    # breakpoint()   # ← uncomment to pause here in the debugger.
                     #   Then type:  p price  (print a value),  n  (next line),  c  (continue)
    if rate < 0 or rate > 1:
        raise ValueError("rate must be between 0 and 1")
    return price * (1 - rate)

final = calculate_discount(100.0, 0.15)
print(final)
```

**Debugging Workflow**

- `breakpoint()` drops you into `pdb` (or an alternative debugger).
- Inspect any local variable with `p varname`.
- Step with `n` (next) / `s` (step into).
- Continue execution with `c`.
- For richer UX many teams use `ipdb` or IDE-integrated debuggers.

### Profiling Bottlenecks

When code is correct but slow, `cProfile` (or the higher-level `py-spy` / `scalene`) shows exactly which functions consume the most time. A typical invocation is:

`cProfile` is run from the command line. In Colab, first save a small script, then run the command with `!`:

```python
%%writefile my_script.py
def slow_square_sum(n):
    total = 0
    for i in range(n):
        total += i * i
    return total

def main():
    for _ in range(5):
        slow_square_sum(300_000)

if __name__ == "__main__":
    main()
```

```python
!python -m cProfile -s cumulative my_script.py | head -n 15
```

### Production-Ready Project Hygiene

**Project Layout**

- Standard `src/` layout
- Clear separation of package code, tests, and docs
- Easy editable installs with `pip install -e .`

**Quality Tooling**

- **Ruff** / **Black** — lightning-fast formatting & linting
- **Mypy** — static type checking
- Lock files (poetry.lock / uv.lock / requirements.txt) for reproducibility

> 💡 **Reproducibility Rule** Always commit a lock file. It guarantees that every developer, CI runner, and production server installs exactly the same package versions, eliminating “works on my machine” surprises.
