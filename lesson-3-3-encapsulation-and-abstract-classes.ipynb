## Introduction & Core Concepts of Encapsulation & Abstract Classes

Have you ever wondered how Python lets you hide the inner workings of your code so only the important parts are exposed? That is the role of **encapsulation**.

Imagine a TV remote. You do not need to know how the electronics inside work — you simply press a button and the volume changes or the channel switches. Encapsulation works the same way: it keeps certain data and methods private so other parts of the program cannot mess with them directly.

> 💡 **Key insight:** Encapsulation bundles data (attributes) and the methods that operate on that data into a single unit (the class), while controlling how the outside world can read or change that data.

### Public vs Protected vs Private

Python does not have true private members the way some languages do, but it follows strong conventions:

- **Public** — no leading underscore. Freely accessible from anywhere.
- **Protected** — single leading underscore (`_password`). A signal to other developers: “this is intended for internal use; please do not touch it from outside.”
- **Private (name-mangled)** — double leading underscore (`__secret`). Python rewrites the name so it becomes harder to access accidentally.

```python
class UserAccount:
    def __init__(self, username, password):
        self.username = username          # public
        self._password = password         # protected (convention)
        self.__login_attempts = 0         # name-mangled

    def check_password(self, attempt):
        if attempt == self._password:
            self.__login_attempts = 0
            return True
        self.__login_attempts += 1
        return False

    def get_attempts(self):
        return self.__login_attempts


account = UserAccount("alice", "s3cr3t")
print(account.username)           # public — OK
print(account._password)          # protected — possible but discouraged
# print(account.__login_attempts)  # AttributeError
print(account.get_attempts())     # controlled access
```

**What the code demonstrates**

- `username` is public and can be read freely.
- `_password` is protected by convention — still reachable, but a clear warning.
- `__login_attempts` is name-mangled; direct access raises an error.
- A public method provides the only safe way to inspect the private counter.

> 💡 **Mental model:** Think of the class as a remote control. Public methods are the buttons you are allowed to press. Protected and private attributes are the internal circuitry that should stay hidden.

## Architecture & Implementation Deep Dive

For stricter privacy Python uses **name mangling**. When you write `__secret`, the interpreter rewrites the attribute name to `_ClassName__secret`. This makes accidental access from outside the class much harder.

**Name Mangling**

Double underscore triggers automatic rewriting. It is not true privacy, but it prevents most unintended collisions and external access.

**Controlled Access**

When you still need to let users read or modify private data, Python’s `@property` decorator gives you getters and setters that look like normal attributes.

### Properties — Getters & Setters

Think of a property as giving someone a remote control that only lets them change the volume, never the channel. You expose a clean interface while keeping validation and side-effects under your control.

```python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def celsius(self):
        """Getter — read the value"""
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        """Setter — validate before storing"""
        if value < -273.15:
            raise ValueError("Below absolute zero!")
        self._celsius = value

    @property
    def fahrenheit(self):
        """Computed property — no setter needed"""
        return self._celsius * 9/5 + 32


t = Temperature(25)
print(t.celsius)        # 25
print(t.fahrenheit)     # 77.0

t.celsius = 30          # uses the setter
print(t.fahrenheit)     # 86.0

# t.celsius = -300     # would raise ValueError
```

**Step-by-step logic**

- `@property` turns a method into a readable attribute.
- `@celsius.setter` intercepts assignment and runs validation.
- `fahrenheit` is a computed read-only property.
- Client code looks like normal attribute access, yet remains safe.

### Transition to Abstract Classes

Encapsulation protects the internal state of a single class. When you design a whole family of related classes you often need a stronger contract — a blueprint that every subclass must follow. That is where abstract classes come in.

> 💡 **Architecture note:** Use name mangling sparingly (mostly to avoid collisions in complex inheritance). Prefer single-underscore conventions plus properties for everyday encapsulation. Reserve abstract base classes for enforcing a required interface across many implementations.

## Best Practices & Production Integration — Abstract Classes

Imagine you are designing a blueprint for vehicles. You know every vehicle should have a `start_engine` method, but you do not yet know exactly how each concrete vehicle will implement it. Abstract classes let you define these blueprints.

In Python you use the `abc` module (Abstract Base Classes). You mark methods with `@abstractmethod`; any concrete subclass that fails to implement them will raise an error when you try to instantiate it.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        """Every vehicle must implement this"""
        pass

    @abstractmethod
    def stop_engine(self):
        pass

    def honk(self):
        # concrete method shared by all vehicles
        return "Beep!"


class Car(Vehicle):
    def start_engine(self):
        return "Car engine started with key"

    def stop_engine(self):
        return "Car engine stopped"


class ElectricScooter(Vehicle):
    def start_engine(self):
        return "Scooter powered on silently"

    def stop_engine(self):
        return "Scooter powered off"


# v = Vehicle()          # TypeError: Can't instantiate abstract class
car = Car()
print(car.start_engine())
print(car.honk())

scooter = ElectricScooter()
print(scooter.start_engine())
```

**Why this matters in production**

- Guarantees every subclass provides the required methods.
- Documents the expected interface clearly.
- Lets you write polymorphic client code that only depends on the abstract type.
- Fails fast at instantiation time if a method is missing.

### Combining Encapsulation with Abstract Classes

A well-designed hierarchy often uses both techniques together: the abstract base class defines the public contract, while each concrete class encapsulates its own private state and implementation details.

**Encapsulation**

Keeps data safe and tidy. Hides internal representation, exposes only controlled interfaces via properties and public methods.

**Abstract Classes**

Ensure designs stay consistent and flexible. Force subclasses to implement required behavior, enabling reliable polymorphism.

> 💡 **Big takeaway:** Encapsulation keeps your code safe and tidy; abstract classes ensure your designs are consistent and flexible. Master both and you write Python that is not merely functional, but robust, organized, and ready for real-world growth.
