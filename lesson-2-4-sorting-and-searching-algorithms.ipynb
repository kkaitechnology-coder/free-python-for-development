## Introduction & Core Concepts of Sorting & Searching

Imagine you’re organizing a bookshelf. You want to arrange the books alphabetically, and later you’ll want to quickly find a particular title. That’s exactly what sorting and searching algorithms do in Python — they organize data efficiently and make it easy to locate what you need.

Two classic sorting algorithms you’ll encounter are **Merge Sort** and **Quicksort**.

> 💡 **Merge Sort Mental Model** Merge sort works like sorting books by first splitting them into small piles, sorting each pile, and then carefully merging the piles back together in order. It is reliable and consistent, especially for large datasets.

### Why Sorting Matters

- Makes later searching dramatically faster
- Enables efficient algorithms that assume ordered data
- Improves readability and reporting of results
- Is the foundation of many database and indexing techniques

Merge sort always takes roughly the same amount of time regardless of the initial order of the data, which makes its performance predictable.

## Quicksort & Binary Search

Quicksort picks a “pivot” element, then rearranges the list so that everything smaller goes to one side and everything larger goes to the other. It is usually faster than merge sort for most practical cases, but it can slow down if the data is already sorted or nearly sorted.

**Merge Sort**

- Divide-and-conquer by splitting
- Stable and predictable
- Always O(n log n)
- Uses extra memory for merging

**Quicksort**

- Divide-and-conquer around a pivot
- Very fast in practice
- Average O(n log n), worst O(n²)
- Can be done in-place

> 💡 **Binary Search** Once a collection is sorted, binary search lets you find an item by repeatedly looking at the middle element and discarding half of the remaining search space. This is far faster than checking every item one by one.

You keep splitting the search space in half until you find the target or determine it is not present. With hundreds or thousands of items the difference in speed is dramatic.

## Big-O Notation & Practical Python Tools

All these algorithms are measured by **Big-O notation** — a way to estimate how much work an algorithm does as the data grows.

- Merge sort & average Quicksort → **O(n log n)**
- Binary search → **O(log n)**
- Linear search (checking every item) → **O(n)**

> 💡 **Python Helpers** You rarely need to implement these algorithms from scratch. Python’s built-in `sorted()`, list `.sort()`, and the `bisect` module already give you highly optimized sorting and binary search.

### Working Examples

```python
import bisect

# Sorting with the built-in tools
books = ["Dune", "Foundation", "Neuromancer", "Hyperion", "Snow Crash"]
sorted_books = sorted(books)          # returns a new list
print("Sorted:", sorted_books)

books.sort()                          # sorts in place
print("In-place:", books)

# Binary search with bisect
# bisect_left returns the insertion point
idx = bisect.bisect_left(books, "Hyperion")
print("Index of Hyperion:", idx)
print("Found:", books[idx] if idx < len(books) and books[idx] == "Hyperion" else "Not found")

# Linear search for comparison
def linear_search(lst, target):
    for i, item in enumerate(lst):
        if item == target:
            return i
    return -1

print("Linear search:", linear_search(books, "Dune"))
```

**Key Execution Notes**

- `sorted()` returns a new sorted list; the original stays unchanged.
- `list.sort()` sorts the list in place and returns `None`.
- `bisect_left` finds the insertion point in a sorted list (binary search).
- Binary search requires the collection to be sorted first.
- Prefer the built-in tools — they are implemented in C and extremely fast.

Whenever you need to organize or find data quickly in Python, these sorting and searching ideas (and the built-in functions that implement them) are the backbone of fast, responsive applications.
