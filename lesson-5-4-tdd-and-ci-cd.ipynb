## Introduction & Core Concepts of TDD & CI/CD

Let’s dive into two powerful practices that really set professional Python development apart: **Test-Driven Development (TDD)** and **Continuous Integration / Continuous Deployment (CI/CD)**. These might sound like buzzwords, but they are practical techniques that help you write better code and deliver it faster and more reliably.

First, TDD. Imagine you are building a bridge. Would you start pouring concrete before knowing how strong it needs to be? Of course not. In TDD you write a test for a tiny piece of functionality before you write the production code itself.

> 💡 **Key insight:** TDD flips the usual order. Instead of “code first, test later”, you write the test first. The failing test becomes a precise specification of what the code must do.

### Why TDD Matters

- Forces you to think about the interface and expected behavior before implementation.
- Gives you a safety net that runs in seconds after every change.
- Encourages small, focused units of work.
- Produces living documentation: the tests describe how the system is supposed to behave.

> 💡 **Mental model:** The test is the blueprint. The code is the concrete. You never pour concrete without a blueprint that has already been checked.

## The Red-Green-Refactor Cycle

The classic TDD rhythm has three stages:

**1. Red**

Write a test for the next small behavior. Run it — it fails because the code does not exist yet (or does not satisfy the requirement).

**2. Green**

Write the simplest code that makes the test pass. Do not worry about elegance yet.

3. **Refactor** — clean up the code while keeping every test green. Then start the next cycle.

### A Tiny TDD Walk-through

Each step below writes a file and runs pytest, so you can watch the test go red → green → stay green.

````python
%%writefile test_calculator.py
def test_add():
    from calculator import add
    assert add(2, 3) == 5
````

````python
# 🟢 GREEN — the minimal implementation passes
!python -m pytest -q test_calculator.py
````

````python
%%writefile calculator.py
def add(a: float, b: float) -> float:
    """Return the sum of a and b."""
    return a + b
````

````python
# 🔵 REFACTOR — cleaner code, tests still pass
!python -m pytest -q test_calculator.py
````

**Rules of the game**

- Never write production code without a failing test.
- Never write more test than is sufficient to fail.
- Never write more production code than is sufficient to pass.
- Refactor only when the bar is green.

> 💡 **Architecture tip:** Keep the cycles tiny (minutes, not hours). Small steps give rapid feedback and keep the design under control.

## CI/CD — Automating the Safety Net

When you work on a team (or even alone on a serious project), every change must be verified automatically. That is the job of **Continuous Integration** and **Continuous Deployment**.

- **CI** — every push or pull-request triggers an automated pipeline that installs dependencies, runs the test suite, checks coverage, lints the code, and reports success or failure.
- **CD** — once the pipeline is green, the same system can automatically deploy the new version to staging or production.

### Minimal GitHub Actions Workflow

GitHub Actions reads this file from `.github/workflows/ci.yml` in your repository — it runs on GitHub, not in Colab:

```yaml
# .github/workflows/ci.yml
name: CI

on:
  push:
    branches: [main]
  pull_request:
    branches: [main]

jobs:
  test:
    runs-on: ubuntu-latest
    steps:
      - uses: actions/checkout@v4
      - uses: actions/setup-python@v5
        with:
          python-version: "3.12"
      - name: Install dependencies
        run: |
          python -m pip install --upgrade pip
          pip install -r requirements.txt
          pip install pytest pytest-cov
      - name: Run tests with coverage
        run: pytest --cov=src --cov-report=term-missing
      - name: Fail if coverage < 80%
        run: |
          coverage report --fail-under=80
```

**What a solid pipeline usually includes**

- Unit & integration tests
- Coverage threshold
- Linting / type-checking (ruff, mypy)
- Security scanning (optional)
- Build & publish artifacts or deploy

**TDD benefit**

Design stays focused, regressions are caught in seconds, and the test suite becomes living documentation.

**CI/CD benefit**

Every change is verified the same way, on a clean machine, before it can reach users. Confidence replaces hope.

> 💡 **Big takeaway:** TDD keeps your code solid from the ground up by forcing a clear specification before implementation. CI/CD makes sure every change is automatically tested and can be delivered smoothly. Together they turn fragile projects into robust, reliable Python applications and make you a far more confident developer.
