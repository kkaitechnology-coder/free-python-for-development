## Introduction & Core Concepts of Classes, Objects & Constructors

Let’s dive into one of the most important building blocks of Python’s object-oriented programming: **classes**, **objects**, and **constructors**.

Imagine you’re designing a video game and you want to create different types of characters. Instead of writing all the details for every single character from scratch, you can use a class as a kind of blueprint.

> 💡 **Blueprint Analogy** A class defines what properties and behaviors every character should have — health, speed, the ability to jump, and so on. When you use this blueprint to create an individual character, that is called an **object** (or instance). If the class is the recipe, the object is the actual cake you bake.

### Key Ideas

- **Class** — the blueprint or template
- **Object / Instance** — a concrete thing created from the class
- Many objects can be created from the same class
- Each object can have its own unique data while sharing the same structure and behavior

This separation of blueprint and concrete instance is what makes object-oriented code reusable and easier to maintain.

## Constructors & the Role of self

Whenever you create a new object from a class, Python automatically calls a special method named `__init__`. This is the **constructor** — the setup process that runs for every new object.

Think of it as the initialization step for every new character in your game, where you decide their starting health, name, or other properties.

> 💡 **What is self?** Inside methods (including `__init__`) you will see the word `self`. `self` refers to the specific object that is being created or operated on. It is how Python knows which object’s data you are reading or changing.

### How Construction Works

- You call the class like a function: `Character("Aria", 100)`
- Python creates a new empty object
- Python calls `__init__` and passes the new object as `self`
- Inside `__init__` you attach attributes to that object
- The finished object is returned to you

You almost never call `__init__` yourself — Python does it automatically when you create an instance.

## Instance Attributes vs Class Attributes

There is an important difference between **instance attributes** and **class attributes**.

**Instance Attributes**

- Belong to one specific object
- Set inside `__init__` (or later on that object)
- Example: each character’s current health

**Class Attributes**

- Shared by all objects of the class
- Defined directly in the class body
- Example: maximum number of players allowed

> 💡 **Takeaway** Classes are your blueprints, objects are the things you build from them, and constructors are the special methods that set everything up when you create something new.

### Working Example

```python
class Character:
    # Class attribute – shared by all instances
    max_players = 4

    def __init__(self, name, health):
        # Instance attributes – unique to each object
        self.name = name
        self.health = health

    def take_damage(self, amount):
        self.health -= amount
        print(f"{self.name} has {self.health} health left")

# Create two different objects from the same class
hero = Character("Aria", 100)
enemy = Character("Goblin", 30)

print("Hero:", hero.name, hero.health)
print("Enemy:", enemy.name, enemy.health)
print("Max players allowed:", Character.max_players)

hero.take_damage(25)
enemy.take_damage(10)
```

**Key Execution Notes**

- The class defines the shared structure and behavior.
- `__init__` runs automatically and receives `self` plus any arguments you pass.
- Attributes assigned to `self` become instance attributes.
- Attributes defined in the class body are class attributes (shared).
- Methods also receive `self` so they can work with the correct object’s data.

Understanding these concepts is the first step toward building more complex, reusable, and maintainable Python programs.
