## Introduction & Core Concepts of Stacks

Have you ever wondered how your browser’s back button remembers the pages you’ve visited, or how a chat app manages messages as they come and go? Today we’re looking at three essential data structures that make these features possible: **stacks**, **queues**, and **linked lists**.

These structures might sound intimidating at first, but they’re built on simple principles — different ways of organizing and managing collections of data, each with its own special rules.

> 💡 **Stack Mental Model** Imagine a stack of plates in your kitchen. You add clean plates to the top and take plates off the top — never from the bottom. This is called **LIFO** (Last In, First Out).

### Why Stacks Matter

- Browser history / back-button navigation
- Undo / redo functionality in editors
- Function call stacks in programming languages
- Parsing expressions and matching parentheses

In Python the most convenient way to implement a stack is with `collections.deque`, using only one end for both push and pop operations.

## Queues — First In, First Out

Now picture a line at a ticket counter. The first person in line is the first one served. That’s a **queue**, which follows **FIFO** (First In, First Out).

**Stack (LIFO)**

- Add and remove from the same end
- Last item added is first removed
- Great for “most recent” behavior
- Example: undo stack

**Queue (FIFO)**

- Add at one end, remove from the other
- First item added is first removed
- Great for fair ordering
- Example: message or task processing

> 💡 **Implementation Tip** Python’s `collections.deque` is ideal for both structures. For a queue you append to the right and pop from the left (or vice versa). It gives O(1) performance at both ends.

Queues appear in print-job scheduling, message brokers, breadth-first search, and any system that must process items in the order they arrived.

## Linked Lists & Practical Takeaways

Unlike stacks and queues, a **linked list** is like a chain of train cars where each car knows about the next one. Each node stores a value and a reference (pointer) to the following node.

Linked lists are less common in everyday Python because the built-in list already covers most needs, but they are excellent for understanding how data can be connected and traversed, and they form the foundation of many advanced structures.

> 💡 **Big Takeaway** Understanding stacks, queues, and linked lists gives you powerful tools for managing data — whether you’re building undo buttons, processing messages, or learning how larger systems work under the hood.

### Working Examples

```python
from collections import deque

# ---- Stack (LIFO) ----
stack = deque()
stack.append("page1")      # push
stack.append("page2")
stack.append("page3")
print("Back button:", stack.pop())   # → page3

# ---- Queue (FIFO) ----
queue = deque()
queue.append("customer1")  # enqueue
queue.append("customer2")
queue.append("customer3")
print("Now serving:", queue.popleft())  # → customer1

# ---- Simple Linked List Node ----
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

head = Node("A")
head.next = Node("B")
head.next.next = Node("C")

# Traverse
current = head
values = []
while current:
    values.append(current.value)
    current = current.next
print("Linked list:", values)
```

**Key Execution Notes**

- Use `deque.append` + `deque.pop` for a stack (LIFO).
- Use `deque.append` + `deque.popleft` for a queue (FIFO).
- A linked-list node holds a value and a reference to the next node.
- Traversing a linked list follows the `next` pointers until `None`.
- Choose the structure that matches the access pattern you need.

With these three structures in your toolkit you can model many real-world ordering and linking problems cleanly and efficiently.
