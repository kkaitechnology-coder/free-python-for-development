## Introduction & Core Concepts of Logging & Debugging

Let’s imagine you are building a Python program that analyzes sales data. Everything works perfectly when you test with a small file, but when you run it on the full dataset something goes wrong. The program crashes, or worse, produces strange results. How do you figure out what happened?

This is where **logging** and **debugging** come in — they are your detectives for tracking down problems in your code.

> 💡 **Key insight:** `print()` is fine for quick experiments, but it is not suitable for production. Logging gives you levels, timestamps, destinations, and the ability to turn detail up or down without touching the source code.

### Two Complementary Tools

- **Logging** — a permanent, structured diary of what the program did while it ran. Ideal for post-mortem analysis and monitoring.
- **Debugging (pdb)** — an interactive pause button. You stop execution, inspect variables, and step through code line by line.

> 💡 **Mental model:** Logging is the flight recorder that keeps running in the background. Debugging is the ability to freeze the plane mid-flight, walk around the cockpit, and examine every instrument.

## The logging Module — Levels, Handlers & Configuration

The standard-library `logging` module records messages at different importance levels: `DEBUG`, `INFO`, `WARNING`, `ERROR` and `CRITICAL`. You can send those messages to the console, to a file, or to both, and you can change the verbosity without rewriting code.

**Levels (lowest → highest)**

DEBUG → INFO → WARNING → ERROR → CRITICAL. A logger set to WARNING will ignore DEBUG and INFO messages.

**Handlers & Formatters**

Handlers decide where the message goes (stream, file, email…). Formatters decide how it looks (timestamp, level name, message…).

### Practical Logging Setup

```python
import logging

# Configure once at the start of the program
logging.basicConfig(
    level=logging.DEBUG,
    force=True,   # replace the notebook's default logging setup
    format="%(asctime)s [%(levelname)s] %(name)s: %(message)s",
    handlers=[
        logging.StreamHandler(),                    # console
        logging.FileHandler("sales.log", mode="a")  # file
    ]
)

logger = logging.getLogger("sales_analyzer")

def process_record(record):
    logger.debug("Processing record id=%s", record.get("id"))
    if "amount" not in record:
        logger.error("Missing 'amount' in record %s", record)
        return
    logger.info("Record %s processed, amount=%.2f",
                record["id"], record["amount"])

process_record({"id": 101, "amount": 49.99})
process_record({"id": 102})  # missing amount
```

**Why this is better than print**

- Timestamps and levels are automatic.
- You can filter by level at runtime.
- Messages go to multiple destinations.
- No need to delete debug prints before shipping.

> 💡 **Architecture tip:** Configure logging as early as possible (preferably in the application entry point). Use named loggers (`getLogger(__name__)`) so each module can be controlled independently.

## Interactive Debugging with pdb & Best Practices

When logs show that something is wrong but you still cannot see why, you need to pause the program and inspect it live. Python’s built-in debugger **pdb** lets you set breakpoints, step through code, and examine variables.

### Using pdb

```python
def calculate_discount(price, percent):
    # Insert a breakpoint
    # breakpoint()   # ← uncomment to pause here in the debugger.
                     #   Then type:  p percent  (print a value),  n  (next line),  c  (continue)
    if percent < 0 or percent > 100:
        raise ValueError("percent must be 0-100")
    return price * (1 - percent / 100)

# When the program hits breakpoint() you get an interactive prompt:
# (Pdb) p price          → 100
# (Pdb) p percent        → 15
# (Pdb) n                → next line
# (Pdb) c                → continue

result = calculate_discount(100, 15)
print("Final price:", result)
```

**Essential pdb commands**

- `n` (next) — execute current line
- `s` (step) — step into a function
- `c` (continue) — run until next breakpoint
- `p expr` — print the value of an expression
- `l` — list source around the current line
- `q` — quit the debugger

**Logging habits**

Log at the appropriate level. Include context (ids, counts). Never log secrets. Prefer structured logging (JSON) in production services.

**Debugging habits**

Reproduce the bug first. Use `breakpoint()` or IDE debuggers. Avoid leaving breakpoints in committed code. Combine with good tests so you catch regressions early.

> 💡 **Big takeaway:** Logging gives you a continuous, reviewable record of what your program is doing; debugging tools like pdb let you dig deep and investigate problems interactively. Mastering both turns mysterious crashes into solvable puzzles and makes you a far more effective Python developer.
