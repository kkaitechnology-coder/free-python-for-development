## 1. Introduction & Core Concepts of Decorators & Design Patterns

Let’s dive into one of the most fascinating areas of object-oriented programming in Python: decorators and design patterns. Imagine you’re building a house. The structure, like the walls and roof, represents your core code. But sometimes, you want to add features—maybe a security system or some smart lighting—without tearing down or rewriting the house itself. This is where decorators and design patterns come in—they help you add new behaviors or solve common problems, all while keeping your code clean and flexible.

Let’s start with decorators. In Python, a decorator is a special type of function that allows you to modify or enhance other functions or methods without changing their actual code. Think of it like putting a protective cover on your phone—it adds extra functionality, like shock resistance, without altering the phone itself. For example, if you have several functions and you want to log when each one gets called, you can write a logging decorator.

### A logging decorator

One decorator adds logging to any function — without touching the function’s own code:

```python
import functools


def log_calls(func):
    """Print every call to `func`, with its arguments and result."""
    @functools.wraps(func)            # keep the original name and docstring
    def wrapper(*args, **kwargs):
        print(f"→ calling {func.__name__}{args}")
        result = func(*args, **kwargs)
        print(f"← {func.__name__} returned {result!r}")
        return result
    return wrapper


@log_calls
def add(a, b):
    return a + b


@log_calls
def greet(name):
    return f"Hello, {name}!"


add(2, 3)
greet("Asha")
```

Decorators can also take arguments. This one measures how long a function takes:

```python
import time


def timed(label):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            start = time.perf_counter()
            try:
                return func(*args, **kwargs)
            finally:
                print(f"[{label}] {func.__name__} took {time.perf_counter() - start:.4f}s")
        return wrapper
    return decorator


@timed("reports")
def build_report(rows):
    return sum(i * i for i in range(rows))


build_report(1_000_000)
```

## 2. Architecture & Implementation Deep Dive

Instead of adding logging code to every function, you just “decorate” each one, and the decorator handles the logging behind the scenes. This keeps your code tidy and avoids repetition.

Now, let’s talk about design patterns. These are tried-and-true solutions to common programming problems. Three important patterns you’ll see in Python are the Singleton, Factory, and Observer patterns. The Singleton pattern ensures that a class has only one instance throughout your program—much like there’s only one president in a country at any time. For example, you might use Singleton for a database connection, so you don’t accidentally open multiple connections.

The Factory pattern is like a car factory: you ask for a car, and depending on your request, the factory gives you a sedan, an SUV, or a truck.

### Singleton — one shared instance

Every “new” `DatabaseConnection` is the same object, so the app never opens a second connection:

```python
class DatabaseConnection:
    _instance = None

    def __new__(cls, url="sqlite:///app.db"):
        if cls._instance is None:
            print(f"Opening connection to {url}")
            cls._instance = super().__new__(cls)
            cls._instance.url = url
        return cls._instance


db1 = DatabaseConnection()
db2 = DatabaseConnection()
print("Same object?", db1 is db2)
```

### Factory — ask for a type, get the right object

The caller names what it wants; the factory decides which class to build:

```python
class Sedan:
    wheels = 4
    def describe(self):
        return "Sedan: comfortable city car"


class SUV:
    wheels = 4
    def describe(self):
        return "SUV: room for the whole family"


class Truck:
    wheels = 6
    def describe(self):
        return "Truck: carries heavy loads"


def vehicle_factory(kind: str):
    vehicles = {"sedan": Sedan, "suv": SUV, "truck": Truck}
    try:
        return vehicles[kind.lower()]()
    except KeyError:
        raise ValueError(f"Unknown vehicle type: {kind!r}") from None


for kind in ["sedan", "SUV", "truck"]:
    v = vehicle_factory(kind)
    print(v.describe(), f"({v.wheels} wheels)")
```

## 3. Best Practices & Production Integration

In programming, the Factory pattern lets you create objects without specifying the exact class of object that will be created. This is especially useful when your code needs to be flexible or handle different types of objects.

Finally, the Observer pattern is all about communication. Imagine a group chat where everyone gets notified when someone sends a message. With the Observer pattern, objects can subscribe to updates from another object, so they’re automatically informed when something changes.

The key takeaway? Decorators and design patterns are powerful tools that help you write smarter, more maintainable Python code. By mastering these, you’ll be well-equipped to handle complexity and change in your projects with elegance and ease.

### Observer — subscribers are notified automatically

Like a group chat: everyone who subscribed gets each new message.

```python
class GroupChat:
    def __init__(self, name):
        self.name = name
        self._subscribers = []

    def subscribe(self, callback):
        self._subscribers.append(callback)

    def send(self, sender, message):
        for notify in self._subscribers:
            notify(self.name, sender, message)


def phone_notification(owner):
    def notify(chat, sender, message):
        if sender != owner:
            print(f"📱 {owner}'s phone — {chat}: {sender} says {message!r}")
    return notify


family = GroupChat("Family")
for person in ["Asha", "Ravi", "Meera"]:
    family.subscribe(phone_notification(person))

family.send("Asha", "Dinner at 8?")
```
