## Introduction & Core Concepts of Exceptions & Custom Errors

Let’s talk about **exceptions** and **custom errors** in Python. Imagine you are baking a cake and halfway through you realize you are out of eggs. Instead of throwing the entire recipe away, you handle the situation — maybe by substituting ingredients or running to the store. That is exactly what exceptions are for: they let your program deal with unexpected situations without crashing.

When something goes wrong — dividing by zero, opening a missing file, or calling a method on `None` — Python **raises** an exception. If nothing catches it, the program stops and prints a traceback.

> 💡 **Key insight:** An exception is an object that carries information about what went wrong. Raising it interrupts normal flow; catching it lets you recover or report the problem cleanly.

### Common Built-in Exceptions

- `ZeroDivisionError` — division by zero
- `FileNotFoundError` — file or directory does not exist
- `TypeError` — operation applied to an inappropriate type
- `ValueError` — right type but inappropriate value
- `KeyError` / `IndexError` — missing dictionary key or list index

> 💡 **Mental model:** Think of normal code as a straight road. An exception is a roadblock. With a try/except you can take a detour, log the problem, or show a friendly message instead of letting the whole trip end.

## try / except / finally — Graceful Error Handling

You wrap risky code inside a `try` block. If an exception occurs, Python jumps to the matching `except` block where you decide how to react. An optional `finally` block always runs — perfect for cleanup such as closing files or releasing locks — whether an error happened or not.

**try**

Contains the code that might fail. Execution continues normally if nothing goes wrong.

**except / finally**

`except` handles specific (or all) exceptions. `finally` guarantees cleanup code runs.

### Practical Example

```python
def read_config(path):
    try:
        with open(path, encoding="utf-8") as f:
            return f.read()
    except FileNotFoundError:
        print(f"Config file '{path}' not found. Using defaults.")
        return "{}"
    except PermissionError:
        print(f"No permission to read '{path}'.")
        return None
    finally:
        print("Finished attempt to load config.")

# Demo
print(read_config("missing.json"))
print("---")
print(read_config("/root/secret.json"))  # may raise PermissionError
```

**Best practices**

- Catch the most specific exception first.
- Avoid bare `except:` — it hides bugs.
- Use `finally` (or a context manager) for cleanup.
- Log the exception or re-raise it if you cannot handle it meaningfully.

> 💡 **Architecture tip:** Prefer context managers (`with`) over manual try/finally for resources. They are cleaner and less error-prone.

## Custom Exceptions — Clear Domain Errors

In larger projects you often need errors that speak the language of your domain. For a banking app you might want to signal “withdrawal exceeds balance” instead of a generic `ValueError`. You create a custom exception by subclassing `Exception` (or a more specific built-in).

### Defining and Raising a Custom Error

```python
class InsufficientFundsError(Exception):
    """Raised when a withdrawal would make the balance negative."""
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        message = (
            f"Cannot withdraw ${amount:.2f}. "
            f"Available balance is only ${balance:.2f}."
        )
        super().__init__(message)


class BankAccount:
    def __init__(self, owner, balance=0.0):
        self.owner = owner
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFundsError(self.balance, amount)
        self.balance -= amount
        return self.balance


account = BankAccount("Alice", 100.0)

try:
    account.withdraw(150)
except InsufficientFundsError as err:
    print("Error:", err)
    print("Balance was:", err.balance)
```

**Why custom exceptions help**

- Callers can catch exactly the error they care about.
- Error messages stay consistent and informative.
- You can attach extra data (balance, amount, error codes).
- Stack traces become easier to diagnose in logs.

**Hierarchy tip**

Create a base domain exception (`BankingError`) and let specific errors inherit from it. Callers can then catch the whole family or a single leaf.

**When to raise**

Raise early when preconditions fail. Catch late, close to the boundary (API handler, CLI, UI) where you can turn the error into a user-friendly response.

> 💡 **Big takeaway:** Exceptions are your safety nets. Built-in ones cover common failures; custom ones make domain problems explicit. Combined with disciplined try/except/finally (or context managers) they keep programs running smoothly and give clear feedback when things go wrong — a hallmark of professional, reliable software.
