## Introduction & Core Concepts of Inheritance & Polymorphism

Let’s dive into two core ideas in object-oriented programming: **inheritance** and **polymorphism**. These mechanisms let us build flexible, reusable hierarchies of classes instead of writing the same logic over and over.

Think of inheritance like a family tree. Imagine you have a base class called **Animal**. You want to create new classes—like **Dog** or **Cat**—that already possess all the basic features of Animal (eating, sleeping, breathing) but can also add their own unique behaviors (barking, meowing, fetching).

> 💡 **Key insight:** Inheritance lets child classes reuse the parent’s attributes and methods without rewriting them. The child can extend or override behavior while remaining fully compatible with the parent’s interface.

### Single Inheritance in Practice

In Python the syntax is straightforward: place the parent class name inside parentheses after the child class name.

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        return f"{self.name} is eating."

    def sleep(self):
        return f"{self.name} is sleeping."


class Dog(Animal):          # Dog inherits from Animal
    def bark(self):
        return f"{self.name} says woof!"


class Cat(Animal):          # Cat also inherits from Animal
    def meow(self):
        return f"{self.name} says meow!"


# Usage
rex = Dog("Rex")
print(rex.eat())            # inherited
print(rex.bark())           # Dog-specific

luna = Cat("Luna")
print(luna.sleep())         # inherited
print(luna.meow())          # Cat-specific
```

**What happens**

- `Dog` and `Cat` automatically receive `__init__`, `eat` and `sleep`.
- They only need to define the new behavior that is unique to them.
- You can still treat a `Dog` instance as an `Animal` when needed.

### Multiple Inheritance

Python goes one step further and supports **multiple inheritance**. A class can inherit from more than one parent, combining features from several sources.

For example, you might have a `Swimmer` class and a `Flyer` class. A `Duck` can inherit from both, gaining the ability to swim and to fly without duplicating any code.

> 💡 **Mental model:** Inheritance is “is-a” relationship. A Dog is an Animal. A Duck is a Swimmer and is a Flyer. This relationship is checked at runtime with `isinstance()` and is fundamental to polymorphic design.

## Architecture & Implementation Deep Dive

Multiple inheritance is powerful, but it introduces a classic problem: what happens when two parent classes define a method with the same name? Which version should the child use?

Python solves this with the **Method Resolution Order (MRO)**—a deterministic algorithm that decides the exact sequence in which base classes are searched for a method or attribute.

### Multiple Inheritance + MRO Example

```python
class Swimmer:
    def move(self):
        return "Swimming"


class Flyer:
    def move(self):
        return "Flying"


class Duck(Swimmer, Flyer):
    pass


d = Duck()
print(d.move())                 # Which move() wins?

# Inspect the MRO
print(Duck.__mro__)
print(Duck.mro())
```

**Step-by-step logic**

- Python looks for `move` first on `Duck` itself — not found.
- It then follows the MRO: `Swimmer` appears before `Flyer`.
- Therefore `Swimmer.move` is selected.
- Changing the order to `class Duck(Flyer, Swimmer)` would reverse the result.

**C3 Linearization**

Python uses the C3 linearization algorithm. It guarantees a consistent, monotonic order that respects the order of base classes you declare and the MRO of each parent.

**Practical tip**

Always inspect `ClassName.__mro__` or `ClassName.mro()` when debugging unexpected method resolution in diamond hierarchies or multiple-inheritance designs.

### Diamond Problem (Classic Trap)

When two parents share a common grand-parent, the diamond shape appears. Python’s MRO still produces a single linear order so that the shared grand-parent method is called only once.

> 💡 **Architecture takeaway:** Prefer composition over deep multiple inheritance when the relationships become hard to reason about. Use multiple inheritance mainly for mixin-style behavior (logging, serialization, comparison helpers) rather than for core domain modeling.

## Best Practices & Production Integration — Polymorphism

**Polymorphism** literally means “many forms.” In Python it means you can write code that works with objects of different classes as long as those objects support the expected interface (methods and attributes).

Python’s most common style of polymorphism is **duck typing**: “If it looks like a duck and quacks like a duck, it’s a duck.” You do not care about the concrete class; you only care that the object has a `fly()` method (or whatever method you need).

### Duck Typing in Action

```python
class Airplane:
    def fly(self):
        return "Airplane soaring through clouds"


class Bird:
    def fly(self):
        return "Bird flapping its wings"


class Superman:
    def fly(self):
        return "Superman flying faster than a speeding bullet"


def let_it_fly(thing):
    # Works with any object that has a fly() method
    print(thing.fly())


# Polymorphic calls — same function, different types
let_it_fly(Airplane())
let_it_fly(Bird())
let_it_fly(Superman())
```

**Why this is powerful**

- No common base class is required.
- New classes can be added later without changing `let_it_fly`.
- The function is open for extension, closed for modification (Open/Closed Principle).

### Combining Inheritance + Polymorphism

When you do share a common base, you can still override methods and rely on polymorphism for clean client code.

```python
class Animal:
    def speak(self):
        raise NotImplementedError("Subclasses must implement speak()")


class Dog(Animal):
    def speak(self):
        return "Woof!"


class Cat(Animal):
    def speak(self):
        return "Meow!"


def animal_chorus(animals):
    for a in animals:
        print(a.speak())      # polymorphic call


animal_chorus([Dog(), Cat(), Dog()])
```

**Production habits**

- Prefer duck typing for loose coupling.
- Use abstract base classes (`abc` module) when you need stronger contracts.
- Keep inheritance hierarchies shallow (2–3 levels).
- Document expected interfaces clearly.

> 💡 **Big takeaway:** Inheritance and polymorphism together give your programs a superpower—flexible, reusable, and maintainable design. Inheritance shares implementation; polymorphism lets client code treat different objects uniformly through a common interface or duck-typed protocol. Master both and you write code that is easier to build, extend, and understand.
