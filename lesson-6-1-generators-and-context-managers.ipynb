## Introduction & Core Concepts of Generators & Context Managers

Let’s dive into two powerful Python concepts that make your code more efficient and readable: **generators** and **context managers**.

Imagine you are working with a massive dataset — a log file with millions of entries. Loading the entire file into memory is simply not practical. Generators solve this problem by producing items one at a time, only when you ask for them, using the `yield` keyword.

> 💡 **Key insight:** A generator is a lazy iterator. It does not compute or store the whole sequence up front; it suspends execution and resumes later, handing you the next value just in time.

### Conveyor-Belt Mental Model

Think of a factory conveyor belt. Instead of dumping every finished product onto the floor at once, the belt delivers each item exactly when the next station is ready. That is how a generator works: it yields one value, pauses, and waits for the next request.

- Constant (or very low) memory usage even for huge data streams.
- Natural fit for pipelines, log processing, and infinite sequences.
- Composable with the rich tools in the `itertools` module.

> 💡 **Mental model:** Ordinary functions return once and are finished. Generator functions yield many times and remember where they left off.

## Generators in Depth

You create a generator simply by writing a function that contains at least one `yield` statement. Calling the function returns a generator object; each call to `next()` (or a `for` loop) resumes the function until the next `yield`.

**Generator function**

Uses `yield`. Returns a generator iterator when called. State is preserved between yields.

**itertools**

A treasure chest of ready-made generators: `chain`, `islice`, `groupby`, `cycle`, and many more for building data pipelines.

### Practical Generator Examples

```python
def count_up_to(n):
    """Yield numbers from 1 to n."""
    i = 1
    while i <= n:
        yield i
        i += 1

# Consume the generator
for num in count_up_to(3):
    print(num)

# Generator expression (like a list comprehension but lazy)
squares = (x * x for x in range(5))
print(list(squares))

# Reading a huge file line by line without loading it all
def read_large_file(path):
    with open(path, encoding="utf-8") as f:
        for line in f:          # file object is already a generator
            yield line.rstrip()

# Example usage (safe for multi-GB logs)
# for line in read_large_file("app.log"):
#     if "ERROR" in line:
#         print(line)
```

**Key points**

- `yield` pauses the function and returns a value.
- The next `next()` call continues right after the yield.
- When the function ends, `StopIteration` is raised automatically.
- Generator expressions use parentheses instead of square brackets.

> 💡 **Architecture tip:** Prefer generators whenever you process streams, large files, or potentially infinite sequences. They keep memory usage flat and compose beautifully with `itertools`.

## Context Managers — Safe Resource Handling

You have almost certainly written `with open("file.txt") as f:`. That `with` statement is powered by a **context manager**. Its job is to guarantee that a resource is properly acquired and later released — even if an exception occurs in the middle.

It is like renting a car: you use it for a while, then return it. You never have to remember the cleanup steps yourself; the protocol handles them.

### Writing Your Own Context Manager

```python
from contextlib import contextmanager
import time

@contextmanager
def timer(label="block"):
    """Context manager that measures elapsed time."""
    start = time.perf_counter()
    try:
        yield                # control returns to the with-block
    finally:
        elapsed = time.perf_counter() - start
        print(f"{label} took {elapsed:.3f}s")

# Usage
with timer("data processing"):
    total = sum(range(10_000_000))

# Classic class-based context manager
class ManagedFile:
    def __init__(self, path, mode="r"):
        self.path = path
        self.mode = mode

    def __enter__(self):
        self.file = open(self.path, self.mode, encoding="utf-8")
        return self.file

    def __exit__(self, exc_type, exc_val, exc_tb):
        self.file.close()
        return False          # do not suppress exceptions

with ManagedFile("demo.txt", "w") as f:
    f.write("Hello from a context manager!\n")
```

**Two ways to build them**

- **Class-based** — implement `__enter__` and `__exit__`.
- **Generator-based** — decorate a generator with `@contextmanager`; code before `yield` is setup, code after is teardown.

**Generators shine when…**

You process large or infinite data streams and want constant memory usage and clean pipeline composition.

**Context managers shine when…**

You acquire any resource (file, lock, connection, temporary directory) that must be released reliably.

> 💡 **Big takeaway:** Generators let you handle big data efficiently by producing values on demand. Context managers make resource management safe and readable. Master both and you will write cleaner, safer, and more memory-efficient Python in real-world applications.
