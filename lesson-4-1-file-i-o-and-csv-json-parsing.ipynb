## Introduction & Core Concepts of File I/O & CSV/JSON Parsing

Imagine you are working on a project where you need to save user data for later, or you want to read information from a file and do something useful with it. That is exactly what **file input and output (File I/O)** is for.

In Python, the gateway to any file is the built-in `open()` function. You can open a file for reading, writing, or appending. The second argument tells Python the mode you need.

> 💡 **Key insight:** Always prefer the `with` statement when working with files. It guarantees the file is closed even if an exception occurs, preventing resource leaks.

### Basic File Reading & Writing

```python
# Writing text to a file
with open("data.txt", "w", encoding="utf-8") as f:
    f.write("Hello, Python!\n")
    f.write("File I/O is easy.\n")

# Reading the whole file
with open("data.txt", "r", encoding="utf-8") as f:
    content = f.read()
    print(content)

# Reading line by line
with open("data.txt", "r", encoding="utf-8") as f:
    for line in f:
        print("Line:", line.strip())
```

**Common modes**

- `"r"` — read (default)
- `"w"` — write (overwrites)
- `"a"` — append
- `"x"` — exclusive create
- `"b"` — binary mode
- `"t"` — text mode (default)

> 💡 **Mental model:** Think of `open()` as unlocking a door to a filing cabinet. The mode decides whether you can only look inside, replace everything, or add new pages. The `with` block is the polite habit of locking the cabinet when you leave.

## Architecture & Implementation Deep Dive — CSV

When the data is structured — rows and columns like a spreadsheet — the most common format is **CSV** (Comma-Separated Values). Python’s built-in `csv` module turns those rows into lists or dictionaries with almost no effort.

**csv.reader**

Returns each row as a plain list of strings. Simple and fast when column order is fixed.

**csv.DictReader**

Uses the first row as field names and returns each subsequent row as a dictionary. Ideal for readable, column-name-based access.

### Reading a CSV File

```python
# Create the sales.csv file this example reads
from pathlib import Path

Path("sales.csv").write_text(
    "date,product,amount\n"
    "2024-01-15,Widget,120.50\n"
    "2024-01-16,Gadget,89.00\n",
    encoding="utf-8",
)
```

```python
import csv

# Assume sales.csv contains:
# date,product,amount
# 2024-01-15,Widget,120.50
# 2024-01-16,Gadget,89.00

with open("sales.csv", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"{row['date']}: {row['product']} → ${row['amount']}")

# Writing a CSV
with open("output.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "score"])
    writer.writeheader()
    writer.writerow({"name": "Alice", "score": 95})
    writer.writerow({"name": "Bob", "score": 87})
```

```python
# See what was written to output.csv
print(Path("output.csv").read_text(encoding="utf-8"))
```

**Why DictReader is preferred**

- Code stays readable when columns are added or reordered.
- You access data by name instead of fragile numeric indexes.
- `newline=""` prevents extra blank rows on Windows.
- Always specify `encoding="utf-8"` for consistent results.

> 💡 **Architecture tip:** Treat CSV as a transport format, not a database. For large or frequently queried data, load the CSV once into memory or a real database and work from there.

## Best Practices & Production Integration — JSON & pathlib

**JSON** is the lingua franca of web APIs and configuration files. Python’s `json` module converts between JSON strings and native Python dictionaries/lists with two main functions: `json.load` / `json.dump` (for files) and `json.loads` / `json.dumps` (for strings).

### Working with JSON

```python
import json

# Python → JSON file
user = {
    "name": "Alice",
    "roles": ["admin", "editor"],
    "active": True
}

with open("user.json", "w", encoding="utf-8") as f:
    json.dump(user, f, indent=2)

# JSON file → Python
with open("user.json", encoding="utf-8") as f:
    data = json.load(f)
    print(data["name"])
    print(data["roles"])
```

**Remember**

- `dump`/`load` work with file objects.
- `dumps`/`loads` work with strings.
- `indent=2` makes files human-readable.
- JSON only supports a limited set of types (no sets, no complex numbers).

### Modern Path Handling with pathlib

The `pathlib` module gives you an object-oriented way to build and inspect file paths that works the same on Windows, macOS and Linux.

```python
from pathlib import Path

# Build a path safely
data_dir = Path("data") / "reports"
data_dir.mkdir(parents=True, exist_ok=True)

report = data_dir / "sales_2024.csv"
print(report.exists())
print(report.suffix)          # .csv
print(report.stem)            # sales_2024

# Read text in one line
text = Path("data.txt").read_text(encoding="utf-8")
print(text[:20])
```

**Production habits**

- Prefer `pathlib.Path` over string concatenation.
- Always open text files with an explicit encoding.
- Use `with` for every file operation.
- Validate file existence and size before heavy processing.
